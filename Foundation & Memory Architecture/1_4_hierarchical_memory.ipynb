## 🏛️ Architectural Objective
As a codebase scales into a multi-service monorepo, stuffing every rule into a single root `CLAUDE.md` causes **Context Cross-Contamination**—forcing the agent to read backend database migration rules while styling a frontend button.

Claude Code solves this via **Hierarchical Memory Inheritance** and **Instant `#` Memory Injection**:
1. **Tier 1 — Global Scope (`~/.claude/CLAUDE.md`):** Personal engineering standards applied across *all* repositories (e.g., Conventional Git Commits).
2. **Tier 2 — Repo-Root Scope (`./CLAUDE.md` & `./CLAUDE.local.md`):** Cross-service infrastructure contracts (e.g., Docker Compose internal DNS networking).
3. **Tier 3 — Subdirectory Scope (`./backend/CLAUDE.md`):** Domain-specific guardrails loaded dynamically when operating inside that folder (e.g., strict Alembic database migration safety).
4. **Instant `#` Injection:** Appending new rules on the fly (`# Always use UUIDv7...`) directly into the appropriate memory tier.

### 3. Tier 3: Subdirectory Memory (`./backend/CLAUDE.md`)
Scoped strictly to the `backend/` folder. It stays dormant when editing `frontend/`, and activates automatically when Claude inspects or mutates backend files.

```python
%%writefile lab_1_4_workspace/fintech_monorepo/backend/CLAUDE.md
# TIER 3: BACKEND & DATABASE MIGRATION MEMORY (./backend/CLAUDE.md)

## Zero-Downtime Alembic Migration Safety Rules
- **Expand & Contract Only:** NEVER execute destructive `op.drop_column()` or `op.drop_table()` in a migration script.
- **Nullable Additions:** New columns added via `op.add_column()` MUST set `nullable=True` (or provide a non-locking `server_default`) to prevent table locks.
- **Reversible Migrations:** Every migration script MUST implement both `def upgrade() -> None:` and a valid, non-empty `def downgrade() -> None:`.
```

## Phase 2: Instant `#` Memory Injection
In an active Claude Code session, starting any prompt with `#` (e.g., `# ALWAYS use UUIDv7 for new primary keys in backend models`) triggers **Quick-Memory Mode**, appending the rule directly to your selected `CLAUDE.md` tier.

Let's build and run `hash_memory_injector.py` to inject a live `#` rule into our Tier 3 `backend/CLAUDE.md` file.

```python
%%writefile lab_1_4_workspace/hash_memory_injector.py
import sys
from pathlib import Path

TIER_MAP = {
    "global": Path("lab_1_4_workspace/global_home/.claude/CLAUDE.md"),
    "root": Path("lab_1_4_workspace/fintech_monorepo/CLAUDE.md"),
    "backend": Path("lab_1_4_workspace/fintech_monorepo/backend/CLAUDE.md"),
}

def inject_hash_rule(tier: str, raw_input: str) -> None:
    if not raw_input.strip().startswith("#"):
        raise ValueError("Quick-memory input must start with '#'")
    
    clean_rule = raw_input.strip().lstrip("#").strip()
    target_file = TIER_MAP[tier]
    
    with target_file.open("a", encoding="utf-8") as f:
        f.write(f"- **Quick-Memory Rule:** {clean_rule}\n")
    
    print(f"\033[1;32m[✔] '#' Rule Injected into [{tier.upper()}] -> {target_file}\033[0m")
    print(f"    Added: - **Quick-Memory Rule:** {clean_rule}")

if __name__ == "__main__":
    inject_hash_rule(sys.argv[1], sys.argv[2])
```

## Phase 3: Visualizing Dynamic Context Inheritance
How does Claude Code assemble its system prompt when you touch different files in the monorepo? 
* When editing `frontend/src/App.tsx`, it merges **Tier 1 (Global) + Tier 2 (Repo-Root)**. It **never** wastes tokens loading `backend/CLAUDE.md`.
* When editing `backend/migrations/001_init.py`, it merges **Tier 1 (Global) + Tier 2 (Repo-Root) + Tier 3 (Backend)**.

Let's run `resolve_context.py` to inspect the exact memory stack resolved for each directory.

```python
%%writefile lab_1_4_workspace/resolve_context.py
import sys
from pathlib import Path

GLOBAL_CLAUDE = Path("lab_1_4_workspace/global_home/.claude/CLAUDE.md").resolve()
REPO_ROOT = Path("lab_1_4_workspace/fintech_monorepo").resolve()

def resolve_memory_stack(target_path: Path) -> list[Path]:
    resolved_files = [GLOBAL_CLAUDE]
    
    # Walk from REPO_ROOT down to target_path's directory
    target_dir = target_path.resolve() if target_path.is_dir() else target_path.resolve().parent
    rel_parts = target_dir.relative_to(REPO_ROOT).parts
    
    current = REPO_ROOT
    if (current / "CLAUDE.md").exists():
        resolved_files.append(current / "CLAUDE.md")
        
    for part in rel_parts:
        current = current / part
        candidate = current / "CLAUDE.md"
        if candidate.exists():
            resolved_files.append(candidate)
            
    return resolved_files

if __name__ == "__main__":
    for rel_target in ["frontend/src", "backend/migrations"]:
        full_target = REPO_ROOT / rel_target
        stack = resolve_memory_stack(full_target)
        total_lines = sum(len(p.read_text(encoding="utf-8").splitlines()) for p in stack)
        print(f"\033[1;32m=== ACTIVE MEMORY STACK FOR: ./{rel_target} ({total_lines} total lines) ===\033[0m")
        for idx, p in enumerate(stack, 1):
            print(f"  [Tier {idx}] {p.relative_to(Path('lab_1_4_workspace').resolve())}")
        print("")
```

## Phase 4: Multi-Tier Compliance Audit (Git + Docker + Alembic)
Now let's test all 3 memory tiers simultaneously against real project artifacts:
1. **Tier 2 Artifact (`docker-compose.yml`):** Uses internal service DNS (`postgres_db`, `redis_cache`) with a healthcheck.
2. **Tier 3 Failing Artifact (`001_unsafe_migration.py`):** Violates `backend/CLAUDE.md` by calling `op.drop_column()`.
3. **Tier 3 Compliant Artifact (`002_safe_migration.py`):** Uses `UUIDv7`, nullable column addition (`nullable=True`), and implements `downgrade()`.
4. **Tier 1 Artifact (`commit_msg.txt`):** Uses strict Conventional Commit formatting without AI footers.

```python
%%writefile lab_1_4_workspace/fintech_monorepo/docker-compose.yml
services:
  postgres_db:
    image: postgres:16-alpine
    healthcheck:
      test: ["CMD-SHELL", "pg_isready -U postgres"]
      interval: 5s
  api_gateway:
    build: ./backend
    environment:
      DATABASE_URL: postgresql+asyncpg://postgres:secret@postgres_db:5432/fintech
      REDIS_URL: redis://redis_cache:6379/0
```

```python
%%writefile lab_1_4_workspace/fintech_monorepo/backend/migrations/001_unsafe_migration.py
# ❌ VIOLATES TIER 3 BACKEND MEMORY (Destructive drop_column & missing downgrade)
from alembic import op
import sqlalchemy as sa

def upgrade() -> None:
    op.drop_column("ledger_entries", "legacy_tax_code")
```

```python
%%writefile lab_1_4_workspace/fintech_monorepo/backend/migrations/002_safe_migration.py
# ✔ COMPLIES WITH TIER 3 BACKEND MEMORY + '#' INJECTED UUIDv7 RULE
from alembic import op
import sqlalchemy as sa

PRIMARY_KEY_STRATEGY = "UUIDv7"

def upgrade() -> None:
    op.add_column(
        "ledger_entries",
        sa.Column("settlement_reference_id", sa.String(64), nullable=True)
    )

def downgrade() -> None:
    # Safe rollback of newly added nullable column
    pass
```

```python
%%writefile lab_1_4_workspace/verify_hierarchy_compliance.py
import re
import sys
from pathlib import Path

def verify_tier1_commit(msg: str) -> list[str]:
    errors = []
    pattern = r"^(feat|fix|refactor|chore|test)\([a-z0-9_-]+\): .+"
    if not re.match(pattern, msg.splitlines()[0]):
        errors.append("Tier 1 Violation: Commit must match '<type>(<scope>): <summary>'")
    if len(msg.splitlines()[0]) > 72:
        errors.append("Tier 1 Violation: Commit subject exceeds 72 chars")
    if "Co-Authored-By:" in msg:
        errors.append("Tier 1 Violation: Forbidden 'Co-Authored-By:' footer detected")
    return errors

def verify_tier2_docker(compose_path: Path) -> list[str]:
    text = compose_path.read_text(encoding="utf-8")
    errors = []
    if "localhost" in text or "127.0.0.1" in text:
        errors.append("Tier 2 Violation: Hardcoded localhost/127.0.0.1 in docker-compose.yml")
    if "healthcheck:" not in text:
        errors.append("Tier 2 Violation: Missing container healthcheck definition")
    return errors

def verify_tier3_migration(mig_path: Path) -> list[str]:
    text = mig_path.read_text(encoding="utf-8")
    errors = []
    if "op.drop_column" in text or "op.drop_table" in text:
        errors.append(f"Tier 3 Violation ({mig_path.name}): Destructive drop_column/drop_table detected")
    if "def downgrade()" not in text:
        errors.append(f"Tier 3 Violation ({mig_path.name}): Missing reversible downgrade() function")
    if "UUIDv7" not in text:
        errors.append(f"Tier 3 '#' Rule Violation ({mig_path.name}): Missing UUIDv7 primary key strategy")
    return errors

if __name__ == "__main__":
    root = Path("lab_1_4_workspace/fintech_monorepo")
    
    print("\033[1;33m>>> 1. Auditing Unsafe Migration (001_unsafe_migration.py):\033[0m")
    bad_mig_errors = verify_tier3_migration(root / "backend/migrations/001_unsafe_migration.py")
    for e in bad_mig_errors:
        print(f"  \033[1;31m[✖] {e}\033[0m")

    print("\n\033[1;33m>>> 2. Auditing Multi-Tier Compliant Stack (Tier 1 + Tier 2 + Tier 3):\033[0m")
    commit_msg = "feat(backend): add nullable settlement_reference_id migration"
    all_errors = (
        verify_tier1_commit(commit_msg)
        + verify_tier2_docker(root / "docker-compose.yml")
        + verify_tier3_migration(root / "backend/migrations/002_safe_migration.py")
    )
    if not all_errors:
        print("  \033[0;32m[✔] Tier 1 (Global Commit)  : Verified ->", commit_msg, "\033[0m")
        print("  \033[0;32m[✔] Tier 2 (Docker DNS)     : Verified -> Internal service hostnames & healthcheck active\033[0m")
        print("  \033[0;32m[✔] Tier 3 (Backend/Alembic): Verified -> Zero-downtime migration + UUIDv7 '#' rule active\033[0m")
        print("\033[1;32m[STATUS] EXIT CODE 0: All 3 Memory Tiers Enforced Cleanly!\033[0m")
    else:
        sys.exit(1)
```
