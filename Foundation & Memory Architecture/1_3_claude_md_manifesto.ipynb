## Phase 2: Engineering the `<150 Line` `CLAUDE.md` Manifesto
A production `CLAUDE.md` is **not** a README for humans—it is a deterministic instruction set for an autonomous compiler. 

Below, we write our trimmed, high-signal `CLAUDE.md` under 65 lines using **Contrastive Rules** (`NEVER` vs `ALWAYS`).

```python
%%writefile lab_1_3_workspace/fintech_monorepo/CLAUDE.md
# FINTECH MONOREPO — ARCHITECTURAL MEMORY MANIFESTO

## 1. Build, Lint & Verification Commands
- **Backend Tests:** `pytest services/api_gateway -q --tb=short`
- **Backend Lint/Format:** `ruff check --fix services/api_gateway && ruff format services/api_gateway`
- **Frontend Typecheck:** `cd apps/web_portal && npx tsc --noEmit`
- **Rule:** NEVER mark a task complete until `pytest` returns POSIX Exit Code `0`.

## 2. Repository Topology
- `services/api_gateway/app/` -> FastAPI (Python 3.11+, Pydantic v2, SQLAlchemy 2.0 Async)
- `apps/web_portal/src/app/`  -> Next.js 15 App Router (TypeScript 5, Tailwind CSS)

## 3. Strict Backend Contracts (FastAPI / Pydantic v2 / SQLAlchemy 2.0)
- **Pydantic v2 Only:**
  - ALWAYS use `model_config = ConfigDict(from_attributes=True)`.
  - ALWAYS serialize models with `.model_dump()` and validate with `.model_validate()`.
  - NEVER use deprecated Pydantic v1 `.dict()`, `.parse_obj()`, `@validator`, or `class Config: orm_mode = True`.
- **SQLAlchemy 2.0 Async Only:**
  - ALWAYS use `AsyncSession`, `Mapped[]`, `mapped_column()`, and `await session.execute(select(Model))`.
  - NEVER use legacy SQLAlchemy 1.x synchronous `session.query(Model)` or untyped `Column()`.
- **Typing & Async:**
  - All FastAPI route handlers touching I/O MUST be `async def` with explicit return type annotations.

## 4. Strict Frontend Contracts (Next.js App Router & Tailwind)
- **Routing & Rendering:**
  - ALWAYS use Next.js App Router (`src/app/**/page.tsx`) and React Server Components (RSC) by default.
  - Add `'use client'` ONLY when browser hooks (`useState`, `useEffect`) or event listeners are required.
  - NEVER create files in `pages/` or use legacy `getServerSideProps` / `getStaticProps`.
- **Styling:**
  - ALWAYS style components exclusively with Tailwind CSS utility classes (`className="..."`).
  - NEVER use inline `style={{...}}` objects or external `.css` module files.
```

## Phase 3: Unguided Legacy Hallucination vs. Manifesto-Guided Execution
Let's see what happens when a developer asks for a Ledger Transaction service **without** `CLAUDE.md` vs. **with** `CLAUDE.md`.

First, we write `legacy_hallucination.py`—representing the exact Pydantic v1 and SQLAlchemy 1.x anti-patterns generated by an unguided LLM.

```python
%%writefile lab_1_3_workspace/fintech_monorepo/services/api_gateway/app/legacy_hallucination.py
# ❌ UNGUIDED LLM OUTPUT (Violates Pydantic v2 & SQLAlchemy 2.0 Contracts)
from pydantic import BaseModel

class TransactionCreate(BaseModel):
    account_id: str
    amount: float

    # VIOLATION 1: Legacy Pydantic v1 inner Config class
    class Config:
        orm_mode = True

def create_transaction_sync(db_session, payload: TransactionCreate):
    # VIOLATION 2: Deprecated Pydantic v1 .dict() call
    raw_data = payload.dict()
    
    # VIOLATION 3: Legacy SQLAlchemy 1.x synchronous .query() call
    existing = db_session.query(TransactionCreate).filter_by(account_id=raw_data["account_id"]).first()
    return raw_data
```

Now let's write `ledger_service.py` and `TransactionCard.tsx`—representing the code generated when Claude Code is governed by our `CLAUDE.md` manifesto.

```python
%%writefile lab_1_3_workspace/fintech_monorepo/services/api_gateway/app/ledger_service.py
# ✔ CLAUDE.md COMPLIANT OUTPUT (Strict Pydantic v2 + SQLAlchemy 2.0 Async Patterns)
from pydantic import BaseModel, ConfigDict, Field
from typing import Dict, Any, List

class TransactionSchema(BaseModel):
    model_config = ConfigDict(from_attributes=True)
    
    account_id: str = Field(..., min_length=3)
    amount: float = Field(..., gt=0.0)
    currency: str = Field(default="USD")

class AsyncSelectSimulator:
    """Lightweight async DB session simulator following SQLAlchemy 2.0 select() semantics."""
    def __init__(self):
        self._store: List[Dict[str, Any]] = []

    async def execute_insert(self, schema: TransactionSchema) -> Dict[str, Any]:
        # Compliant Pydantic v2 serialization via .model_dump()
        record = schema.model_dump()
        self._store.append(record)
        return record

async def process_ledger_entry(session: AsyncSelectSimulator, raw_input: Dict[str, Any]) -> Dict[str, Any]:
    # Compliant Pydantic v2 validation via .model_validate()
    validated = TransactionSchema.model_validate(raw_input)
    return await session.execute_insert(validated)
```

```python
%%writefile lab_1_3_workspace/fintech_monorepo/apps/web_portal/src/app/TransactionCard.tsx
// ✔ CLAUDE.md COMPLIANT NEXT.JS APP ROUTER SERVER COMPONENT (Tailwind Only)
import React from 'react';

interface TransactionCardProps {
  accountId: string;
  amount: number;
  currency: string;
}

export default function TransactionCard({ accountId, amount, currency }: TransactionCardProps) {
  return (
    <div className="rounded-lg border border-emerald-500/30 bg-black p-4 text-emerald-400 shadow-md">
      <p className="text-xs uppercase tracking-wider text-neutral-400">Account: {accountId}</p>
      <p className="mt-1 text-2xl font-mono font-bold">
        {currency} {amount.toFixed(2)}
      </p>
    </div>
  );
}
```

## Phase 4: Automated `CLAUDE.md` Contract Verification & Pytest Guardrails
As Technical Architects, we pair `CLAUDE.md` with an automated contract scanner (`verify_claude_contracts.py`) and unit tests (`test_ledger_service.py`). 

Let's create the contract scanner and test suite, then run them against both the legacy file and the compliant codebase.

```python
%%writefile lab_1_3_workspace/fintech_monorepo/verify_claude_contracts.py
import sys
from pathlib import Path

FORBIDDEN_PATTERNS = {
    ".dict(": "Forbidden Pydantic v1 '.dict()' detected -> Use '.model_dump()'",
    "orm_mode": "Forbidden Pydantic v1 'orm_mode' detected -> Use 'ConfigDict(from_attributes=True)'",
    ".query(": "Forbidden SQLAlchemy 1.x '.query()' detected -> Use 'await session.execute(select(...))'",
    "getServerSideProps": "Forbidden Next.js Pages Router 'getServerSideProps' detected -> Use App Router RSC",
    "style={{": "Forbidden inline React 'style={{}}' detected -> Use Tailwind CSS utility classes",
}

def audit_file(filepath: Path) -> list[str]:
    content = filepath.read_text(encoding="utf-8")
    violations = []
    for pattern, reason in FORBIDDEN_PATTERNS.items():
        if pattern in content:
            violations.append(f"{filepath.name}: {reason}")
    return violations

if __name__ == "__main__":
    target_files = [Path(p) for p in sys.argv[1:]]
    all_violations = []
    for f in target_files:
        all_violations.extend(audit_file(f))
    
    if all_violations:
        print("\033[1;31m[✖] CLAUDE.md CONTRACT VIOLATIONS DETECTED:\033[0m")
        for v in all_violations:
            print(f"  - {v}")
        sys.exit(1)
    else:
        print("\033[1;32m[✔] 100% CLAUDE.md ARCHITECTURAL CONTRACT COMPLIANCE!\033[0m")
        sys.exit(0)
```

```python
%%writefile lab_1_3_workspace/fintech_monorepo/services/api_gateway/app/test_ledger_service.py
import asyncio
from ledger_service import AsyncSelectSimulator, process_ledger_entry

def test_pydantic_v2_and_async_flow():
    session = AsyncSelectSimulator()
    payload = {"account_id": "ACC-9001", "amount": 250.75, "currency": "USD"}
    result = asyncio.run(process_ledger_entry(session, payload))
    assert result["account_id"] == "ACC-9001"
    assert result["amount"] == 250.75
```
